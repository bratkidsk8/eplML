```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

epl= pd.read_csv('/Users/saulmartinezs/Documents/VS/epl_env/Dataset/epl_final.csv')

sns.set_style('dark')
sns.set_palette('hls',8)

datos_columnas=epl[[
    'FullTimeHomeGoals',
    'FullTimeAwayGoals',
    'FullTimeResult',
    'HalfTimeResult',
    'HomeShots',
    'AwayShots',
    'HomeShotsOnTarget',
    'AwayShotsOnTarget',
    'HomeCorners',
    'AwayCorners',
    'HomeFouls',
    'AwayFouls',
    'HomeYellowCards',
    'AwayYellowCards',
    'HomeRedCards',
    'AwayRedCards'
    ]]

for variable in datos_columnas:
    plt.figure(figsize=(8,6))
    sns.boxplot(x=epl[variable])
    plt.tight_layout()
    plt.show()
```
